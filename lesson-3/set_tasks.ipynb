```python
# Setlarning birlashmasi. Ikkita set berilgan. Ikkala setdagi barcha unique elementlarni o‘z ichiga olgan
#yangi set yarating.

st = {'a',1,3,4}
st2 = {'b','c',4,2,1}

st_sum = st.union(st2)
print(st_sum)
```

```python
A = st - st.difference(st2)
B = st2 - st2.difference(st)
print(set(tuple(A) + tuple(B)))
"""Setlarning kesishmasi. Ikkita set berilgan. Har ikkala set uchun umumiy bo‘lgan elementlardan yangi
set yarating.
"""

# oson usuli intersection() ni ishlatsa buladi

print(st.intersection(st2))
```

```python
""" Element qo‘shish. Set va element berilgan. Agar element setda hali mavjud bo‘lmasa, uni qo‘shing.
"""

d_set = {1,2,3,4,5}
n = 6
if n not in d_set:
    d_set.add(n)
    print(d_set)
else:
    print("setda bu raqam mavjud")
```

```python
j_set  = {0,2,1,2,3,4,5,8,4}

ns = set()
for i in j_set:
    if i % 2 == 0:
        ns.add(i)
print(ns)
# filterda chiqargandaaa

fil_set = set(filter(lambda x: x % 2 == 0, j_set))
print(fil_set)
""" Juft sonlarni filtrlash. Butun sonlardan iborat set berilgan. Faqat juft sonlardan tashkil topgan yangi
set yarating.
"""
```

```python
"""Toq sonlarni filtrlash. Butun sonlardan iborat set berilgan. Faqat toq sonlardan tashkil topgan yangi
set yarating.
"""
j_set  = {0,2,1,2,3,4,5,8,4}

toq_set = set(filter(lambda x: x % 2 != 0, j_set))
print(toq_set)

```

```python
#Oraliqdan set yaratish. Ko‘rsatilgan oraliqdagi sonlardan set yarating (masalan, 1 dan 10 gacha).

oraliq_set = set()

for i in range(0,10):
    oraliq_set.add(i)
print(oraliq_set)
```

```python
"""Birlashtirish va takrorlarni olib tashlash. Ikkita list berilgan. Ularni birlashtiradigan va takroriy
elementlarni olib tashlaydigan yangi set yarating.
"""
ls1 = [1,2,3,4,5]
ls2 = [1,'c',3,'a',5]

sew = set(ls1 + ls2)
print(sew)
```

```python
set01 = {1,2,3,4,5}
set02 = {'hi','hello','ok'}

nlset = set01 & set02
nlset1 = set01.intersection(set02) #umumiy elemnt oladi bumasa set() qiymat chiqadi
print(bool(nlset))
print(bool(nlset1))

print(set01.isdisjoint(set02)) #umumiy elelmti yuq bulsa true
"""O‘zaro kesishmaydigan setlarni tekshirish. Ikkita set berilgan. Ularning umumiy elementi
yo‘qligini tekshiring.."""
```
